```python

from pyspark.sql import functions as F
```

# Leer tabla filtrada

```python
YYYY_MM = "2025-08"

anio = int(YYYY_MM[:4])
mes = int(YYYY_MM[5:7])

NAME_LH = "LH_BRONCE_ADVANCED"
NAME_SCHEMA = "NYCTLC"
NAME_TABLE = "yellow_tripdata"

df = (
    spark.table(f"{NAME_LH}.{NAME_SCHEMA}.{NAME_TABLE}")
    .filter(
        (F.col("pickup_year") == anio) &
        (F.col("pickup_month") == mes)
    )
)

display(df)
```

# Leer directamente los archivos físicos de esa carpeta 

```python
YYYY_MM = "2025-08"

anio = YYYY_MM[:4]
mes = YYYY_MM[5:7]

path = f"Tables/NYCTLC/{NAME_TABLE}/pickup_year={anio}/pickup_month={mes}"

df = spark.read.format("delta").load(path)

display(df)
```

# Leer la partición como parte de la tabla Delta 

```python
YYYY_MM = "2025-08"

anio = YYYY_MM[:4]
mes = YYYY_MM[5:7]

path = "Tables/NYCTLC/yellow_tripdata"

df = (
    spark.read
    .format("delta")
    .load(path)
    .filter(
        (F.col("pickup_year") == anio) &
        (F.col("pickup_month") == mes)
    )
)

display(df)
```
