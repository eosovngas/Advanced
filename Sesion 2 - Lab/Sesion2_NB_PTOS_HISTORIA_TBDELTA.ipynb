# Aperturar json operationParameters & operationMetrics

```python
from pyspark.sql import functions as F

df_history = spark.sql("""
    DESCRIBE HISTORY LH_BRONCE_ADVANCED.NYCTLC.yellow_tripdata
""").select(
    "version",
    "timestamp",
    "operation",

    # Operation Parameters
    F.col("operationParameters.mode").alias("mode"),
    F.col("operationParameters.partitionBy").alias("partitionBy"),
    F.col("operationParameters.predicate").alias("predicate"),

    # Operation Metrics
    F.col("operationMetrics.numFiles")
        .cast("long")
        .alias("numFiles"),

    F.col("operationMetrics.numRemovedFiles")
        .cast("long")
        .alias("numRemovedFiles"),

    F.col("operationMetrics.numRemovedBytes")
        .cast("long")
        .alias("numRemovedBytes"),

    F.col("operationMetrics.numCopiedRows")
        .cast("long")
        .alias("numCopiedRows"),

    F.col("operationMetrics.numAddedChangeFiles")
        .cast("long")
        .alias("numAddedChangeFiles"),

    F.col("operationMetrics.numDeletedRows")
        .cast("long")
        .alias("numDeletedRows"),

    F.col("operationMetrics.numOutputRows")
        .cast("long")
        .alias("numOutputRows"),

    F.col("operationMetrics.numOutputBytes")
        .cast("long")
        .alias("numOutputBytes")
)

display(df_history)
```

# Diccionario de campos

| Métrica               |      Valor | Interpretación                      |
| --------------------- | ---------: | ----------------------------------- |
| `numFiles`            |          2 | Archivos generados por la operación |
| `numRemovedFiles`     |          0 | Archivos eliminados                 |
| `numRemovedBytes`     |          0 | Bytes eliminados                    |
| `numCopiedRows`       |          0 | Filas copiadas                      |
| `numAddedChangeFiles` |          0 | Archivos de cambios agregados       |
| `numDeletedRows`      |          0 | Filas eliminadas                    |
| `numOutputRows`       |  3.574.074 | Filas generadas como resultado      |
| `numOutputBytes`      | 76.023.752 | Bytes generados como resultado      |
